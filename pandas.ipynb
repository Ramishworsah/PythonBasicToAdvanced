```python
import pandas as pd
import numpy as np
data={
    'name':['Tom','Jerry','Mickey','Donald'],
    'age':[20,21,22,23],
    'city':['New York','Los Angeles','Chicago','Houston']  ,
    'salary':[50000,60000,70000,80000]
}

pd.DataFrame(data)
```

```python
data_list=[
    ['Tom',20,'New York',50000],['Jerry',21,'Los Angeles',60000],
    ['Mickey',22,'Chicago',70000],['Donald',23,'Houston',80000]]

df=pd.DataFrame(data_list,columns=['name','age','city','salary'])
```

Missing data  and finding null value

```python
data={
    'A': [1, 2, np.nan, 4],
    'B': [5, np.nan, np.nan, 8],
    'C': [9, 10, 11, 12],
    'D': [np.nan, 14, 15, 16]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isnull().sum()
```

removing null values

```python
df.dropna()
```

```python
df.dropna(thresh=3)
```

```python
df.fillna(0)
```

```python
values={
    'A': 0,'B': 100,'C': 200,'D': 400}
df.fillna(value=values)
```

```python
df.fillna(df.mean())
```
