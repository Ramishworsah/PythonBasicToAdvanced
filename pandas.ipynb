mergin and joining concatinations 

```python
import pandas as pd
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4],
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'department': ['HR', 'Finance', 'IT', 'Marketing']})


salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 5],
    'salary': [70000, 80000, 90000, 100000],
    'bonus': [5000, 6000, 7000, 8000]})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees, salaries, on='employee_id')
```

```python
pd.merge(employees,salaries,on='employee_id',how='inner')
```

```python
pd.merge(employees,salaries,on='employee_id',how='outer')
```

grouped and aggregation in pnadas


```python
data={
    'category': ['A', 'B', 'A', 'B', 'C'],
    'value': [10, 20, 30, 40, 50],
    'sales': [100, 200, 300, 400, 500],
    'quantity': [1, 2, 3, 4, 5],
    'store': ['Store1', 'Store2', 'Store1', 'Store2', 'Store3'],
    'date':['2023-01-01', '2023-01-02', '2023-01-03', '2023-01-04', '2023-01-05']
}

df3=pd.DataFrame(data)
```

```python
df3
```

```python
cat=df3.groupby('category')['sales'].sum()
cat
```

```python
st=df3.groupby('store')['sales'].sum()
st
```

```python
df3.groupby(['category','store'])['sales'].sum()
```

#Pivot Table

```python
data3={
    'Date':pd.date_range(start='2023-01-01', periods=20,),
    'Product': ['A', 'B', 'C', 'D', 'E']*4,
    'Region':['North','East','South','West','Central','North','East','South','West','Central','North','East','South','West','Central','North','East','South','West','Central'],
    
    'Sales':np.random.randint(100, 500, size=20),
    'Units': np.random.randint(1, 10, size=20),
    'Rep':['Rep1', 'Rep2', 'Rep3', 'Rep4', 'Rep5', 'Rep1', 'Rep2', 'Rep3', 'Rep4', 'Rep5',
           'Rep1', 'Rep2', 'Rep3', 'Rep4', 'Rep5', 'Rep1', 'Rep2', 'Rep3', 'Rep4', 'Rep5']
}

df4=pd.DataFrame(data3)

df4['Month'] = df4['Date'].dt.month_name()
df4
```

```python
pd.pivot_table(df4,values='Sales', index=['Product', 'Region'], columns='Month', aggfunc='sum', fill_value=0)
```
