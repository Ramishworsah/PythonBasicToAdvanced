```python
def extract_episode(text):
    a=False
    data=""
    for i in text:
        if i==')':
            a=False
            return data
       
        if a==True:
            data+=i
        if i=='(':
            a=True
       
            
```

```python
df['episodes']=df['Title'].apply(extract_episode)
```

```python
df['episodes']=df['episodes'].str.replace(" eps","")
```

```python
def extract_time(text):
    a=False
    data=""
    for i in range(len(text)):
        
        if text[i]==')':
            for j in range(i+1,i+20):
                data+=text[j]
            return data
        
```
